```python
numerical_cols=df[['estimated_income','superannuation_savings','credit_card_balance','bank_loans','bank_deposits','checking_accounts','saving_accounts','foreign_currency_account','business_lending']]
plt.figure(figsize=(20,60))
for i ,col in enumerate(numerical_cols):
   plt.subplot(9,1,i+1)
   sns.histplot(df[col],kde=True)
   plt.title(col)
plt.show()
    
```

```python
numerical_cols=['estimated_income','superannuation_savings','credit_card_balance','bank_loans','bank_deposits','checking_accounts','saving_accounts','foreign_currency_account','business_lending']
correlation_matrix=df[numerical_cols].corr()
plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix,annot=True,cmap='crest')
plt.show()
```

Checking Accounts & Bank Deposits have extremely high correlation (0.84)
1. What it means:

Clients who keep money in checking accounts usually maintain large bank deposits too.

2. Risk Insight:

These customers are low-risk, stable, and have good cash flow.

3. Action:

Bank should target these customers for:

Premium credit cards

Low-risk lending products.
    
 Saving Accounts & Bank Deposits also strongly correlated (0.75)
1. Meaning:

A customer with higher savings also has higher bank deposits.

2. Risk Insight:

Savers are financially disciplined, hence low probability of default.

3. Action:

Cross-sell:

Investment products

Wealth management packages
